# Indian Defence Budget Analytics

A reproducible exploratory data analysis of projected and allocated Indian defence budgets for 2014-15 to 2021-22.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

## 2. Load Dataset

```python
# Keep the CSV in the same folder as this notebook.
csv_path = Path("RS_Session_253_AU1455.A.csv")

if not csv_path.exists():
    raise FileNotFoundError(
        f"Dataset not found: {csv_path}. "
        "Place RS_Session_253_AU1455.A.csv in the notebook folder."
    )

df = pd.read_csv(csv_path)
df.head()
```

## 3. Inspect and Prepare Data

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nMissing values:")
print(df.isna().sum())
```

```python
# Adjust column names here only if the source CSV uses different names.
df.columns = [str(c).strip() for c in df.columns]

df.head()
```

## 4. Derived Metrics

```python
# Expected source columns:
# Year, BE - Projected, BE - Allocated

df["Allocation Gap"] = df["BE - Projected"] - df["BE - Allocated"]
df["Allocation Rate (%)"] = (
    df["BE - Allocated"] / df["BE - Projected"] * 100
)
df["Allocated YoY Growth (%)"] = (
    df["BE - Allocated"].pct_change() * 100
)

analysis_df = df.copy()
analysis_df
```

## 5. Summary Metrics

```python
projected_growth = (
    (df["BE - Projected"].iloc[-1] / df["BE - Projected"].iloc[0]) - 1
) * 100

allocated_growth = (
    (df["BE - Allocated"].iloc[-1] / df["BE - Allocated"].iloc[0]) - 1
) * 100

avg_allocation_rate = df["Allocation Rate (%)"].mean()
avg_gap = df["Allocation Gap"].mean()

highest_allocated_row = df.loc[df["BE - Allocated"].idxmax()]
largest_gap_row = df.loc[df["Allocation Gap"].idxmax()]

print(f"Projected budget growth: {projected_growth:.2f}%")
print(f"Allocated budget growth: {allocated_growth:.2f}%")
print(f"Average allocation rate: {avg_allocation_rate:.2f}%")
print(f"Average projected-vs-allocated gap: ₹{avg_gap:,.2f} crore")
print(
    f"Highest allocated budget: {highest_allocated_row['Year']} — "
    f"₹{highest_allocated_row['BE - Allocated']:,.2f} crore"
)
print(
    f"Largest gap: {largest_gap_row['Year']} — "
    f"₹{largest_gap_row['Allocation Gap']:,.2f} crore"
)
```

## 6. Visualization — Projected vs Allocated Budget

```python
plt.figure(figsize=(10, 6))
plt.plot(df["Year"], df["BE - Projected"], marker="o", label="BE - Projected")
plt.plot(df["Year"], df["BE - Allocated"], marker="o", label="BE - Allocated")
plt.title("Indian Defence Budget: Projected vs Allocated")
plt.xlabel("Financial Year")
plt.ylabel("Budget (₹ crore)")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Visualization — Allocation Rate

```python
plt.figure(figsize=(10, 6))
plt.bar(df["Year"], df["Allocation Rate (%)"])
plt.title("Defence Budget Allocation Rate by Year")
plt.xlabel("Financial Year")
plt.ylabel("Allocation Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 8. Visualization — Allocated Budget YoY Growth

```python
growth = df["Allocated YoY Growth (%)"].copy()

plt.figure(figsize=(10, 6))
plt.plot(df["Year"], growth, marker="o")
plt.axhline(0, linewidth=1)
plt.title("Year-over-Year Growth in Allocated Defence Budget")
plt.xlabel("Financial Year")
plt.ylabel("YoY Growth (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 9. Export Analysis Output

```python
output_path = Path("analysis_output.csv")
analysis_df.to_csv(output_path, index=False)
print(f"Saved: {output_path}")
```

## 10. Key Findings

The supplied analysis reports 58.23% projected-budget growth, 54.30% allocated-budget growth, a 76.62% average allocation rate, an average gap of ₹82,918.18 crore, the highest allocated budget in 2021-22, and the largest gap in 2021-22.
